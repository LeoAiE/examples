```python
import random
import pandas as pd

# Configuration
num_files = 50
rows_per_file = 100

# Sample country codes for generating phone numbers
country_codes = [
    "1", "44", "61", "33", "49", "34", "81", "86", "91", "55",
    "7", "27", "82", "39", "46", "90", "351", "52", "62", "234"
]

def random_phone():
    """Generate a random international phone number (no leading '+')."""
    code = random.choice(country_codes)
    remaining = random.randint(7, 10)
    return code + ''.join(random.choices('0123456789', k=remaining))

def generate_call_records(main_number, rows):
    """Create a DataFrame with `rows` call records for the given main number."""
    records = []
    for _ in range(rows):
        if random.choice([True, False]):  # outgoing call
            from_num = main_number
            to_num = random_phone()
        else:  # incoming call
            from_num = random_phone()
            to_num = main_number
        records.append({"Label": main_number, "From": from_num, "To": to_num})
    return pd.DataFrame(records)

# Ensure all main numbers are unique
main_numbers = set()
while len(main_numbers) < num_files:
    main_numbers.add(random_phone())
main_numbers = list(main_numbers)

# Generate the CSV files
for main_number in main_numbers:
    df = generate_call_records(main_number, rows_per_file)
    filename = f"{main_number}.csv"
    df.to_csv(filename, index=False)

print(f"Created {num_files} CSV files in the current directory.")
```

```python
import os
import random
import string

# -------------------------------  CONFIG  -------------------------------- #
NUM_FILES           = 30      # how many text files to create
ITEMS_PER_FILE      = 20      # contacts per file (comma-separated)

SHARED_TELEGRAM     = "@sharedStar"                    # appears in 10 files
SHARED_TELEGRAM_OCC = 10

SHARED_PHONES       = ["34525267",
                       "+14155500002",
                       "+14155500003"]                # each appears in 20 files
SHARED_PHONE_OCC    = 20

OUTPUT_DIR          = "contact_lists"                 # folder for .txt files
# ------------------------------------------------------------------------- #

# A few real world country codes for random phone generation
COUNTRY_CODES = ["1", "44", "33", "49", "55", "61", "34", "91",
                 "81", "86", "7", "27", "90", "351", "62", "234"]

# ----------------------  helper-generation functions  -------------------- #
def random_phone():
    """Return a random international phone number like +441234567890."""
    code = random.choice(COUNTRY_CODES)
    length = random.randint(7, 10)          # remaining digits
    digits = ''.join(random.choices(string.digits, k=length))
    return f"+{code}{digits}"

def random_username():
    """Return a random Telegram username like @silent_owl3."""
    name_len = random.randint(5, 12)
    name = ''.join(random.choices(string.ascii_lowercase, k=name_len))
    if random.random() < 0.35:              # sprinkle in underscores / digits
        name += random.choice(["_", "", str(random.randint(0, 99))])
    return f"@{name}"
# ------------------------------------------------------------------------- #

# -------------------------  decide shared-item spread  ------------------- #
file_indices = list(range(NUM_FILES))

# Pick which files get the shared Telegram username
telegram_files = set(random.sample(file_indices, SHARED_TELEGRAM_OCC))

# Pick which files each shared phone number lands in
phone_files = {
    ph: set(random.sample(file_indices, SHARED_PHONE_OCC))
    for ph in SHARED_PHONES
}
# ------------------------------------------------------------------------- #

# -----------------------  generate and write files  ---------------------- #
os.makedirs(OUTPUT_DIR, exist_ok=True)

for idx in file_indices:
    contacts = set()

    # add shared telegram (if applicable)
    if idx in telegram_files:
        contacts.add(SHARED_TELEGRAM)

    # add shared phones (if applicable)
    for ph, target_set in phone_files.items():
        if idx in target_set:
            contacts.add(ph)

    # fill up to ITEMS_PER_FILE with random unique contacts
    while len(contacts) < ITEMS_PER_FILE:
        if random.random() < 0.5:
            contacts.add(random_phone())
        else:
            contacts.add(random_username())

    # write to disk
    fname = f"{OUTPUT_DIR}/contacts_{idx+1}.txt"
    with open(fname, "w", encoding="utf-8") as f:
        f.write(", ".join(sorted(contacts)))

print(f"Created {NUM_FILES} files in '{OUTPUT_DIR}/' ✔")
```

```python
import os
import random
import string

# -------------------------------  CONFIG  -------------------------------- #
NUM_FILES           = 30      # how many text files to create
ITEMS_PER_FILE      = 20      # contacts per file (comma-separated)

# Shared items and how often they appear
SHARED_TELEGRAM       = "sharedStar"                     # Telegram username
SHARED_TELEGRAM_OCC   = 10

SHARED_PHONES         = ["767655833",
                         "+14155500002",
                         "+14155500003"]                 # phone numbers
SHARED_PHONE_OCC      = 20     # each phone appears in 20 files

SHARED_HANDLES        = ["alpha@teleHub",
                         "beta@teleHub"]                  # handle style
SHARED_HANDLE_OCC     = 15     # each handle appears in 15 files

OUTPUT_DIR            = "contact_lists"
# ------------------------------------------------------------------------- #

# A few real-world country codes for random phone generation
COUNTRY_CODES = ["1", "44", "33", "49", "55", "61", "34", "91",
                 "81", "86", "7", "27", "90", "351", "62", "234"]

# ----------------------  helper-generation functions  -------------------- #
def random_phone():
    """Random international phone number like +441234567890"""
    code = random.choice(COUNTRY_CODES)
    length = random.randint(7, 10)
    digits = ''.join(random.choices(string.digits, k=length))
    return f"+{code}{digits}"

def random_username():
    """Random Telegram username like @silent_owl3"""
    base = ''.join(random.choices(string.ascii_lowercase, k=random.randint(5, 12)))
    suffix = random.choice(["", "_" + ''.join(random.choices(string.ascii_lowercase, k=3)),
                            str(random.randint(0, 99))])
    return f"@{base}{suffix}"

def random_handle():
    """Random handle like neo@matrixNet"""
    left  = ''.join(random.choices(string.ascii_lowercase, k=random.randint(3, 8)))
    right = ''.join(random.choices(string.ascii_lowercase, k=random.randint(5, 10)))
    return f"{left}@{right}"
# ------------------------------------------------------------------------- #

# --------------------  map shared items to file indices  ----------------- #
file_indices = list(range(NUM_FILES))

# Pick which files get the shared Telegram username
telegram_files = set(random.sample(file_indices, SHARED_TELEGRAM_OCC))

# For each shared phone, pick its files
phone_files = {ph: set(random.sample(file_indices, SHARED_PHONE_OCC))
               for ph in SHARED_PHONES}

# For each shared handle, pick its files
handle_files = {h: set(random.sample(file_indices, SHARED_HANDLE_OCC))
                for h in SHARED_HANDLES}
# ------------------------------------------------------------------------- #

# ------------------------  generate & write files  ----------------------- #
os.makedirs(OUTPUT_DIR, exist_ok=True)

for idx in file_indices:
    contacts = set()

    # add shared telegram username
    if idx in telegram_files:
        contacts.add(SHARED_TELEGRAM)

    # add shared phone numbers
    for ph, target_set in phone_files.items():
        if idx in target_set:
            contacts.add(ph)

    # add shared handles
    for h, target_set in handle_files.items():
        if idx in target_set:
            contacts.add(h)

    # fill the rest with random unique contacts
    while len(contacts) < ITEMS_PER_FILE:
        r = random.random()
        if r < 0.34:
            contacts.add(random_phone())
        elif r < 0.67:
            contacts.add(random_username())
        else:
            contacts.add(random_handle())

    # write contacts to file
    filename = f"{OUTPUT_DIR}/contacts_{idx+1}.txt"
    with open(filename, "w", encoding="utf-8") as f:
        f.write(", ".join(sorted(contacts)))

print(f"✅ Created {NUM_FILES} files in '{OUTPUT_DIR}/'")
```
